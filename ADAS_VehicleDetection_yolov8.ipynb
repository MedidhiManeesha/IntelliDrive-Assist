```python
import os

# Define dataset paths
dataset_path = "/content/drive/MyDrive/VehicleDataset"

# Define subfolders
splits = ["train", "test", "valid"]
total_images = 0

# Loop through train, test, valid
for split in splits:
    image_folder = os.path.join(dataset_path, split, "images")  # Images folder
    num_images = len([f for f in os.listdir(image_folder) if f.endswith(('.jpg', '.png'))])
    print(f"Total images in {split}: {num_images}")
    total_images += num_images

print(f" Total images in dataset: {total_images}")
```

```python
total_labels = 0

for split in splits:
    label_folder = os.path.join(dataset_path, split, "labels")  # Labels folder
    num_labels = len([f for f in os.listdir(label_folder) if f.endswith('.txt')])
    print(f"Total label files in {split}: {num_labels}")
    total_labels += num_labels

print(f" Total label files in dataset: {total_labels}")
```

```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8s.pt")

model.train(data="/content/drive/MyDrive/VehicleDataset/data.yaml",
            epochs=50,
            batch=16,
            imgsz=640,
            save_period=10,  # Save model every 10 epochs
            project="/content/drive/MyDrive/YOLO_vehicle_training",
            name="vehicle_detection")
```

```python
model = YOLO("/content/drive/MyDrive/YOLO_vehicle_training/vehicle_detection2/weights/best.pt")

model.predict(source="/content/video2.mp4", save=True, conf=0.5)
```

```python
import imageio
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

def display_video(video_path):
    reader = imageio.get_reader(video_path)
    fig, ax = plt.subplots(figsize=(8,6))  # Increase figure size

    mov = []
    for frame in reader:
        # frame = np.rot90(frame, k=2)  # Rotate 180° if video is upside down
        img = ax.imshow(frame, animated=True)
        ax.axis('off')
        mov.append([img])

    reader.close()

    anime = animation.ArtistAnimation(fig, mov, interval=50, repeat_delay=1000)
    plt.close()
    return HTML(anime.to_html5_video())

# Path to YOLOv8 output video (update if needed)
video_path = "/content/ultralytics/runs/detect/predict/video2.avi"
display_video(video_path)
```
